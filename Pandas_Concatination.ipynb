```python
import pandas as pd
df1 = pd.DataFrame({
    "Id":[1,2,3],
    "Name":["Tony","Steve","Thor"]
})
                  
df2 = pd.DataFrame({
    "Id":[4,5,6],
    "Name":["Hulk","Spider-Man","Vision"]
})

pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],ignore_index=True)
```

```python
pd.concat([df1,df2],axis=1)
```
